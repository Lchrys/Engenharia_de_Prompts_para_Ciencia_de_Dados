```python
# biblioteca oficial de tokenização da OpenAI:
import tiktoken 

enc = tiktoken.get_encoding("o200k_base") # nome do tokenizador (inclusive para pt-br)

chamados = {
    "A": "Minha internet caiu ontem à noite e só voltou hoje de manhã. É a terceira vez neste mês. Se não resolverem, vou cancelar.",
    "B": "A conexão está sem sinal desde as 19h. Já reiniciei o modem e as luzes continuam vermelhas.",
    "C": "Fui cobrado duas vezes na fatura de setembro. Quero o estorno do valor cobrado em duplicidade.",
    "D": "Gostaria de saber se posso mudar para o plano de 500 Mega sem pagar nova taxa de instalação."}

print("Chamado | Caracteres | Palavras | Tokens | Tokens/Palavra")
for k, t in chamados.items():
    print(f"{k:<7} | {len(t):<10} | {len(t.split()):<8} | {len(enc.encode(t)):<6} | {len(enc.encode(t))/len(t.split()):.2f}")
```
